## 3. Distribution of Life Expectancy

The distribution of life expectancy helps us understand how life expectancy values are spread across the countries in the dataset. A histogram with a density curve is used to identify the most common range and overall distribution.

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["Life_expectancy"],
    bins=20,
    kde=True
)

plt.title("Distribution of Life Expectancy")
plt.xlabel("Life Expectancy (years)")
plt.ylabel("Number of Countries")
plt.tight_layout()
plt.show()
```

```python
life_stats = df["Life_expectancy"].agg([
    "mean",
    "median",
    "min",
    "max",
    "std"
])

life_stats
```

### Interpretation

The descriptive statistics provide an overview of the variation in life expectancy across countries. The difference between the minimum and maximum values indicates substantial variation in life expectancy, which may be associated with differences in health, economic conditions, education, and mortality indicators.

```python
top_10 = df[
    ["Country", "Life_expectancy"]
].sort_values(
    by="Life_expectancy",
    ascending=False
).head(10)

top_10
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_10,
    x="Life_expectancy",
    y="Country"
)

plt.title("Top 10 Countries by Life Expectancy")
plt.xlabel("Life Expectancy (years)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

## 4. Life Expectancy by Region

This section compares average life expectancy across different geographical regions. Regional differences can help identify broader patterns in health outcomes and socioeconomic conditions.

```python
region_life = (
    df.groupby("Region")["Life_expectancy"]
    .mean()
    .sort_values(ascending=False)
)

region_life
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=region_life.values,
    y=region_life.index
)

plt.title("Average Life Expectancy by Region")
plt.xlabel("Average Life Expectancy (years)")
plt.ylabel("Region")
plt.tight_layout()
plt.show()
```

## 5. Life Expectancy by Economic Status

Economic conditions can influence access to healthcare, nutrition, education, sanitation, and other factors that affect population health. This analysis compares average life expectancy between developed and developing countries.

```python
economy_life = (
    df.groupby("Economy_label")["Life_expectancy"]
    .mean()
    .sort_values(ascending=False)
)

economy_life
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=economy_life.index,
    y=economy_life.values
)

plt.title("Average Life Expectancy by Economic Status")
plt.xlabel("Economic Status")
plt.ylabel("Average Life Expectancy (years)")
plt.tight_layout()
plt.show()
```

```python
economy_summary = df.groupby("Economy_label").agg(
    Average_Life_Expectancy=("Life_expectancy", "mean"),
    Number_of_Countries=("Country", "count")
).round(2)

economy_summary
```

## 6. GDP and Life Expectancy

Gross Domestic Product (GDP) per capita is an indicator of a country's economic condition. This analysis investigates whether countries with higher GDP per capita tend to have higher life expectancy.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="GDP_per_capita",
    y="Life_expectancy",
    hue="Economy_label",
    s=80
)

plt.title("GDP per Capita vs Life Expectancy")
plt.xlabel("GDP per Capita")
plt.ylabel("Life Expectancy (years)")
plt.legend(title="Economic Status")
plt.tight_layout()
plt.show()
```

```python
gdp_corr = df["GDP_per_capita"].corr(df["Life_expectancy"])

print("Correlation between GDP per capita and Life expectancy:")
print(round(gdp_corr, 3))
```

## 12. Immunization and Life Expectancy

Immunization coverage is an important indicator of public health. This section examines the relationship between vaccination coverage for Hepatitis B, Polio, and Diphtheria and life expectancy.

```python
immunization_cols = ["Hepatitis_B", "Polio", "Diphtheria"]

correlations = df[immunization_cols + ["Life_expectancy"]].corr()["Life_expectancy"].drop("Life_expectancy")

plt.figure(figsize=(8, 5))

sns.barplot(
    x=correlations.index,
    y=correlations.values
)

plt.title("Correlation of Immunization Indicators with Life Expectancy")
plt.xlabel("Immunization Indicator")
plt.ylabel("Correlation with Life Expectancy")
plt.axhline(0, linewidth=1)
plt.tight_layout()
plt.show()
```

## 13. Correlation Analysis

Correlation analysis is used to identify the strength and direction of relationships between life expectancy and the major health, economic, mortality, and education indicators.

A correlation value close to +1 indicates a strong positive relationship, while a value close to -1 indicates a strong negative relationship. A value close to 0 indicates a weak linear relationship.

```python
# Select numerical variables for correlation analysis

numeric_cols = [
    "Infant_deaths",
    "Under_five_deaths",
    "Adult_mortality",
    "Alcohol_consumption",
    "Hepatitis_B",
    "Measles",
    "BMI",
    "Polio",
    "Diphtheria",
    "Incidents_HIV",
    "GDP_per_capita",
    "Population_mln",
    "Thinness_ten_nineteen_years",
    "Thinness_five_nine_years",
    "Schooling",
    "Life_expectancy"
]

correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of Health, Economic and Social Indicators")
plt.tight_layout()
plt.show()
```

```python
life_corr = (
    correlation_matrix["Life_expectancy"]
    .drop("Life_expectancy")
    .sort_values(ascending=False)
)

life_corr
```

```python
plt.figure(figsize=(10, 7))

sns.barplot(
    x=life_corr.values,
    y=life_corr.index
)

plt.title("Correlation of Variables with Life Expectancy")
plt.xlabel("Correlation")
plt.ylabel("Variable")
plt.axvline(0, linewidth=1)
plt.tight_layout()
plt.show()
```

# 19. Life Expectancy Summary Dashboard

This section presents the key findings from the analysis through a set of summary visualizations.

```python
# ==========================================
# LIFE EXPECTANCY SUMMARY DASHBOARD
# ==========================================

fig, axes = plt.subplots(2, 2, figsize=(16, 11))

# ------------------------------------------
# 1. Life Expectancy Distribution
# ------------------------------------------
sns.histplot(
    df["Life_expectancy"],
    bins=20,
    kde=True,
    ax=axes[0, 0]
)

axes[0, 0].set_title("Distribution of Life Expectancy", fontsize=14)
axes[0, 0].set_xlabel("Life Expectancy (years)")
axes[0, 0].set_ylabel("Number of Countries")


# ------------------------------------------
# 2. Economic Status Comparison
# ------------------------------------------
economy_plot = (
    df.groupby("Economy_label")["Life_expectancy"]
    .mean()
    .sort_values(ascending=False)
)

sns.barplot(
    x=economy_plot.index,
    y=economy_plot.values,
    ax=axes[0, 1]
)

axes[0, 1].set_title(
    "Average Life Expectancy by Economic Status",
    fontsize=14
)

axes[0, 1].set_xlabel("Economic Status")
axes[0, 1].set_ylabel("Life Expectancy (years)")


# ------------------------------------------
# 3. Schooling vs Life Expectancy
# ------------------------------------------
sns.scatterplot(
    data=df,
    x="Schooling",
    y="Life_expectancy",
    hue="Economy_label",
    s=60,
    ax=axes[1, 0]
)

axes[1, 0].set_title(
    "Schooling vs Life Expectancy",
    fontsize=14
)

axes[1, 0].set_xlabel("Schooling")
axes[1, 0].set_ylabel("Life Expectancy (years)")


# ------------------------------------------
# 4. Adult Mortality vs Life Expectancy
# ------------------------------------------
sns.scatterplot(
    data=df,
    x="Adult_mortality",
    y="Life_expectancy",
    hue="Economy_label",
    s=60,
    ax=axes[1, 1]
)

axes[1, 1].set_title(
    "Adult Mortality vs Life Expectancy",
    fontsize=14
)

axes[1, 1].set_xlabel("Adult Mortality")
axes[1, 1].set_ylabel("Life Expectancy (years)")


plt.suptitle(
    "Life Expectancy Analysis — Summary Dashboard",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```
